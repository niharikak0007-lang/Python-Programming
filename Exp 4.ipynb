```python
n = int(input("Enter number: "))
fact = 1

for i in range(1, n+1):
    fact *= i

print("Factorial =", fact)
```

```python
n = int(input("Enter number: "))
temp = n
sum = 0

while temp > 0:
    digit = temp % 10
    sum = sum + (digit * digit * digit)
    temp = temp // 10

if sum == n:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```

```python
n = int(input("Enter terms: "))
a = 0
b = 1

for i in range(n):
    print(a)
    c = a + b
    a = b
    b = c
```

```python
n = int(input("Enter number: "))
count = 0

for i in range(1, n+1):
    if n % i == 0:
        count = count + 1

if count == 2:
    print("Prime Number")
else:
    print("Not a Prime Number")
```

```python
n = int(input("Enter number: "))
temp = n
rev = 0

while temp > 0:
    digit = temp % 10
    rev = rev * 10 + digit
    temp = temp // 10

if rev == n:
    print("Palindrome Number")
else:
    print("Not a Palindrome Number")
```

```python
n = int(input("Enter number: "))
sum = 0

while n > 0:
    digit = n % 10
    sum = sum + digit
    n = n // 10

print("Sum of digits:", sum)
```

```python
count = 0

for i in range(1, 101):
    if i % 5 == 0 or i % 7 == 0:
        print(i)
        count = count + 1

print("Total numbers:", count)
```

```python
s = input("Enter string: ")
result = ""

for ch in s:
    if ch >= 'a' and ch <= 'z':
        result = result + chr(ord(ch) - 32)
    else:
        result = result + ch

print("Uppercase string:", result)
```

```python
for num in range(2, 101):
    count = 0
    for i in range(1, num+1):
        if num % i == 0:
            count = count + 1
    if count == 2:
        print(num)
```
