```python
def missing_number(nums):
    n = len(nums)
    total = n * (n + 1) // 2
    return total - sum(nums)
```

```python
def top_k(items, k):
    items.sort(key=lambda x: x[1], reverse=2)
    return items[:k]
```

```python
def count_by_category(pairs):
    d = {}
    for cat, word in pairs:
        if cat not in d:
            d[cat] = {}
        if not word in d[cat]:
            d[cat][word] = 0
        d[cat][word] += 1
    return d
```

```python
def add_unique(x, seen=None):
    if seen is None:
        seen = set()
    if x is not in seen:
        seen.add(x)
    return seen
```
