```python
x = 9
y = 7

print("Addition:", x + y)
print("Multiplication:", x * y)
print("Division:", x / y)
print("Subtraction:", x - y)
```

```python
r = float(input("Enter radius: "))
area = 3.14159 * r * r
print("Area of circle:", area)
```

```python
x = 4
y = 3

result = (x + y) * (x + y)
print("Result:", result) 
```

```python
a = float(input("Enter base: "))
b = float(input("Enter height: "))

c = (a**2 + b**2) ** 0.5
print("Hypotenuse:", c)
```

```python
p = float(input("Enter principal: "))
r = float(input("Enter rate: "))
t = float(input("Enter time: "))

si = (p * r * t) / 100
print("Simple Interest:", si)
```

```python
a = float(input("Enter side a: "))
b = float(input("Enter side b: "))
c = float(input("Enter side c: "))

s = (a + b + c) / 2
area = (s*(s-a)*(s-b)*(s-c)) ** 0.5
print("Area of triangle:", area)
```

```python
seconds = int(input("Enter seconds: "))

hours = seconds // 3600
minutes = (seconds % 3600) // 60
remaining_seconds = seconds % 60

print("Hours:", hours)
print("Minutes:", minutes)
print("Seconds:", remaining_seconds)
```

```python
a = int(input("Enter a: "))
b = int(input("Enter b: "))

a = a + b
b = a - b
a = a - b

print("After swapping:")
print("a =", a)
print("b =", b)
```

```python
n = int(input("Enter n: "))

sum_n = n * (n + 1) // 2
print("Sum of first", n, "natural numbers:", sum_n)
```

```python
print("A B  A&B  A|B  A^B")
for A in [0,1]:
    for B in [0,1]:
        print(A, B, " ", A&B, "  ", A|B, "  ", A^B)
```

```python
num = int(input("Enter number: "))

print("Left shift by 1:", num << 1)
print("Right shift by 1:", num >> 1)
```

```python
seq = (10, 20, 56, 78, 89)
num = int(input("Enter number: "))

if num in seq:
    print("Number is in sequence")
else:
    print("Number is not in sequence")
```

```python
string = input("Enter a string: ")
char = input("Enter a character: ")

if char in string:
    print("Character found in string")
else:
    print("Character not found in string")
```
