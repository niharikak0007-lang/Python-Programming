```python
s = input("Enter a string: ")
count = 0

for ch in s:
    if ch.isupper():
        count += 1

print("Number of capital letters:", count)
```

```python
s = input("Enter a string: ")
vowels = "aeiouAEIOU"
count = 0

for ch in s:
    if ch in vowels:
        count += 1

print("Number of vowels:", count)
```

```python
sentence = input("Enter a sentence: ")
words = sentence.split()

for word in words:
    print(word)
```

```python
string = input("Enter string: ")
substring = input("Enter substring: ")

count = 0
i = 0

while i <= len(string) - len(substring):
    if string[i:i+len(substring)] == substring:
        count += 1
    i += 1

print(count)
```

```python
s = input("Enter string: ")
s = s.upper()

letters = {}

for ch in s:
    if ch.isalpha():
        if ch in letters:
            letters[ch] += 1
        else:
            letters[ch] = 1

for key in sorted(letters):
    print(letters[key], key)
```

```python
sentence = input("Enter a sentence: ")
words = sentence.split()

unique_words = set(words)

print("Number of unique words:", len(unique_words))
print("Unique words:", unique_words)
```

```python
n = int(input("Enter number of fruits: "))

s1 = set()
s2 = set()

print("Enter fruits for set s1:")
for i in range(n):
    fruit = input()
    s1.add(fruit)

print("Enter fruits for set s2:")
for i in range(n):
    fruit = input()
    s2.add(fruit)

# a) Fruits in both sets
print("Common fruits:", s1 & s2)

# b) Fruits only in s1
print("Only in s1:", s1 - s2)

# c) Count of all fruits
all_fruits = s1 | s2
print("Total fruits count:", len(all_fruits))
```

```python
S1 = {"Red", "yellow", "orange", "blue"}
S2 = {"violet", "blue", "purple"}

print("Union:", S1 | S2)
print("Intersection:", S1 & S2)
print("Difference S1 - S2:", S1 - S2)
print("Difference S2 - S1:", S2 - S1)
```
